## 1. The Many-Time Pad Vulnerability & Crib Dragging

```python
from functools import reduce

def toHex(s):
    return "".join(hex(ord(c))[2:].zfill(2) for c in s)

def toStr(hex_str):
    bytes_arr = [int(hex_str[i:i+2], 16) for i in range(0, len(hex_str), 2)]
    return "".join(chr(b) if 32 <= b <= 126 else '?' for b in bytes_arr)

def Xor(s1, s2):
    res = ""
    for i in range(len(s1)):
        res += format(int(s1[i], 16) ^ int(s2[i], 16), '01x')
    return res

message1 = "steal the secret"
message2 = "the cat the bird"
key      = "supersecretverys"

c1 = Xor(toHex(message1), toHex(key))
c2 = Xor(toHex(message2), toHex(key))
xor_c = Xor(c1, c2)
xor_m = Xor(toHex(message1), toHex(message2))
assert xor_c == xor_m
print("Ciphertexts XOR successfully equals Messages XOR!")
```

## 2. Diffie-Hellman Key Exchange Protocol

```python
# Public parameters
N = 23    # Modulus (prime)
g = 5     # Generator

# Private keys
alice_secret = 6
bob_secret = 15

# Publicly transmitted components
A = pow(g, alice_secret, N)
B = pow(g, bob_secret, N)

# Shared key computation
K_alice = pow(B, alice_secret, N)
K_bob = pow(A, bob_secret, N)

print(f"Alice computed shared key: {K_alice}")
print(f"Bob computed shared key:   {K_bob}")
assert K_alice == K_bob
```

## 3. Order Finding & Shor's Factorization Principle
If $ is the period such that ^r \equiv 1 \pmod N$, and $ is even, then $\gcd(a^{r/2} \pm 1, N)$ yields prime factors of $.

```python
import math

N = 35
a = 8
# Find period r
r = 1
while pow(a, r, N) != 1:
    r += 1
print(f"Period of {a} mod {N} is r = {r}")

if r % 2 == 0 and pow(a, r // 2, N) != N - 1:
    factor1 = math.gcd(pow(a, r // 2) - 1, N)
    factor2 = math.gcd(pow(a, r // 2) + 1, N)
    print(f"Factors of {N} successfully extracted: {factor1} and {factor2}")
    assert factor1 * factor2 == N
```
