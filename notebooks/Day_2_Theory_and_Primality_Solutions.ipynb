## Problem 1: Repunit Sequences & Prime Divisibility
Let (n) = \underbrace{11\dots1}_{n} = \frac{10^n - 1}{9}$. Let  = \{n : p \mid a(n)\}$.
- If  = 2$ or  = 5$,  = \emptyset$ because the last digit of (n)$ is always 1.
- For  \neq 2, 5$,  \mid a(n) \iff 10^n \equiv 1 \pmod p$, which occurs if and only if $\text{ord}_p(10) \mid n$.
Hence  = \{k \cdot \text{ord}_p(10) : k \in \mathbb{Z}^+\}$ is infinite.

```python
def repunit(n):
    return (10**n - 1) // 9

def multiplicative_order(a, p):
    if p == 2 or p == 5:
        return None
    if p == 3:
        return 3 # for repunit: 27 divides 10^3 - 1
    rem = a % p
    order = 1
    while rem != 1:
        rem = (rem * a) % p
        order += 1
        if order > p:
            return None
    return order

for p in [2, 3, 5, 7, 11, 13, 17, 19]:
    ord_p = multiplicative_order(10, p)
    print(f"p = {p:2d}: ord_p(10) = {str(ord_p):4s} | A_p = {f'{ord_p}*k' if ord_p else 'Empty set'}")
```

## Problem 2: Euclid-Euler Theorem on Even Perfect Numbers
A number $ is an even perfect number if and only if  = 2^{n-1}(2^n - 1)$ where ^n - 1$ is a Mersenne prime.

```python
def is_prime(num):
    if num < 2:
        return False
    if num in (2, 3):
        return True
    if num % 2 == 0 or num % 3 == 0:
        return False
    i = 5
    while i * i <= num:
        if num % i == 0 or num % (i + 2) == 0:
            return False
        i += 6
    return True

def sum_of_divisors(num):
    total = 0
    for i in range(1, int(num**0.5) + 1):
        if num % i == 0:
            total += i
            if i * i != num:
                total += num // i
    return total

print("Generating Even Perfect Numbers via Euclid-Euler Formula:")
for n in range(2, 10):
    mersenne = 2**n - 1
    if is_prime(mersenne):
        perfect = 2**(n-1) * mersenne
        is_perf = (sum_of_divisors(perfect) == 2 * perfect)
        print(f"n = {n}: Mersenne prime = {mersenne:3d} -> Perfect number = {perfect:6d} (Verified: {is_perf})")
```

## Problem 3: Fermat Pseudoprimes
A composite integer $ is a Fermat pseudoprime to base $ if ^{n-1} \equiv 1 \pmod n$.

```python
def is_fermat_pseudoprime(n, base=2):
    if is_prime(n):
        return False
    return pow(base, n - 1, n) == 1

pseudoprimes_base2 = [n for n in range(2, 2000) if is_fermat_pseudoprime(n, base=2)]
print("Fermat pseudoprimes to base 2 (< 2000):", pseudoprimes_base2)
# First few: 341 (11 * 31), 561 (3 * 11 * 17), 645 (3 * 5 * 43), 1105 (5 * 13 * 17), 1387 (19 * 73), 1729 (7 * 13 * 19), 1905
```
