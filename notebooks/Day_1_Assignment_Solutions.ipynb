## Problem 1: Extended Euclid's Algorithm
Find  = \gcd(a, b)$ and integers , y$ such that  + by = d$.

```python
def extended_gcd(a, b):
    assert a >= b and b >= 0 and a + b > 0
    if b == 0:
        return (a, 1, 0)
    d, x1, y1 = extended_gcd(b, a % b)
    x = y1
    y = x1 - (a // b) * y1
    assert a % d == 0 and b % d == 0
    assert d == a * x + b * y
    return (d, x, y)

# Test cases
print("extended_gcd(8, 4)  =", extended_gcd(8, 4))
print("extended_gcd(27, 15) =", extended_gcd(27, 15))
```

## Problem 2: Least Common Multiple (LCM)
Calculate $\text{LCM}(a, b) = \frac{a \cdot b}{\gcd(a, b)}$.

```python
def lcm(a, b):
    assert a > 0 and b > 0
    d, _, _ = extended_gcd(max(a, b), min(a, b))
    return (a * b) // d

assert lcm(9, 3) == 9
assert lcm(17, 37) == 629
print("lcm(9, 3)   =", lcm(9, 3))
print("lcm(17, 37) =", lcm(17, 37))
```

## Problem 3: Solving Linear Diophantine Equations
Given  > 0, b > 0, c$, find integers , y$ such that  + by = c$.

```python
def diophantine(a, b, c):
    swapped = False
    if a < b:
        a, b = b, a
        swapped = True
    
    gcd, x0, y0 = extended_gcd(a, b)
    assert c % gcd == 0, "No solution exists because gcd does not divide c"
    factor = c // gcd
    x, y = x0 * factor, y0 * factor
    if swapped:
        a, b = b, a
        x, y = y, x
    assert a * x + b * y == c
    return (x, y)

print("diophantine(34, 51, 170) =", diophantine(34, 51, 170))
print("diophantine(91, 117, 65)  =", diophantine(91, 117, 65))
```

## Problem 4: Modular Division & Multiplicative Inverses
Find $ such that  \equiv b \pmod n$, where $\gcd(a, n) = 1$.

```python
def modInverse(a, n):
    a_mod = a % n
    d, x, y = extended_gcd(n, a_mod)
    assert d == 1, "Inverse does not exist"
    return (y % n + n) % n

def divide(b, a, n):
    assert n > 1 and a > 0
    aInv = modInverse(a, n)
    assert (a * aInv) % n == 1
    return (aInv * b) % n

print("divide(9, 5, 3)     =", divide(9, 5, 3))
print("divide(223, 67, 54) =", divide(223, 67, 54))
```

## Problem 5: Fast Modular Exponentiation
Compute ^e \bmod m$ using binary exponentiation in (\log e)$ steps.

```python
def FastModularExponentiation(b, e, m):
    res = 1
    base = b % m
    exp = e
    while exp > 0:
        if exp % 2 == 1:
            res = (res * base) % m
        base = (base * base) % m
        exp //= 2
    return res

print("FastModularExponentiation(4, 5, 3) =", FastModularExponentiation(4, 5, 3))
assert FastModularExponentiation(4, 5, 3) == pow(4, 5, 3)
```

## Problem 6: Chinese Remainder Theorem (CRT)
Find  \in [0, n_1 n_2)$ such that  \equiv r_1 \pmod{n_1}$ and  \equiv r_2 \pmod{n_2}$ for coprime , n_2$.

```python
def ChineseRemainderTheorem(n1, r1, n2, r2):
    d, x, y = extended_gcd(n1, n2)
    assert d == 1, "Moduli must be coprime"
    r = (r1 * n2 * y + r2 * n1 * x) % (n1 * n2)
    return (r + n1 * n2) % (n1 * n2)

print("CRT(13, 7, 6, 5)   =", ChineseRemainderTheorem(13, 7, 6, 5))   # Expected: 59
print("CRT(28, 20, 19, 3) =", ChineseRemainderTheorem(28, 20, 19, 3)) # Expected: 440
```
